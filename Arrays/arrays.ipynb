## 1. Array

```python
import array
```

```python
# creation

arr1 = array.array( 'i', [10, 20, 30, 40, 50])
arr1
```

#### Access

```python
arr1[2]
```

#### Traversal

```python
i = 0
for element in arr1:
    i += 1
    print(f"{element}", end=", ")
```

#### Insertion

```python
arr1.insert(3, 35)
print(arr1)
```

#### Deletion

```python
arr1.remove(35) # only deletes the first occurence of element, not all 30s only firs occurence of it
print(arr1)
```

#### Updation

```python
arr1[2] = 35
print(arr1)
```

#### Sorting

```python
# creating an array2
arr2 = array.array('i', [60, 70, 80, 90])
print(arr2)
```

```python
arr3 = array.array('i', [4, 5, 6, 7, 8, 9, 2, 1])
print(arr3)
```

```python
sorted_array = array.array('i', sorted(arr3))
print(sorted_array)
```

#### Merging

```python
merged_array = arr1 + arr2
merged_array
```

#### Splitting

```python
ix = len(merged_array) // 2
ix
```

```python
# slicing operations
merged_array[:ix]
```

```python
merged_array[ix:]
```

## 2. Numpy

```python
import numpy as np
```

#### Creation

```python
np1 = np.array([10, 20, 30, 40, 50])
np2 = np.array([50, 60, 70])

print(np1)
print(np2)
np1
```

#### Access

```python
np1[0]
```

#### Traversal

```python
for ele in np1:
    print(ele, end=" ")
```

#### Insertion

```python
np.insert(np1, 3, 35)
```

#### Deletion

```python
np.delete(np1, 3)
```

#### Updation

```python
np1[2] = 35
print(np1)
```

#### Sorting

```python
np3 = np.array([5, 3, 7, 6, 2, 9])
np3
```

```python
np.sort(np3)
```

#### Merging

```python
np1
np2
np3 = np.concatenate((np1, np2))
np3
```

#### Splitting

```python
split1, split2 = np.array_split(np3, 2)
split1
split2
```

#### Vectorization

```python
np1
np1 + 10
np1 * 2
```

```python
np3 = np.array([4, 5])
np4 = np.array([6, 7])

np3 / np4
```

```python
li1 = [4, 5, 6]
li2 = [1, 2, 3]
for i in range(len(li1)):
    print(li1[i] + li2[i])
for i in range(len(li1)):
    print(li1[i] + 10)
```
